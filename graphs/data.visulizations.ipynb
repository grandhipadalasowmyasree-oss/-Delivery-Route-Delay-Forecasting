```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import os

```

```python
df = pd.read_excel(
    r"C:\Users\grand\OneDrive\Desktop\-Delivery-Route-Delay-Forecasting\-Delivery-Route-Delay-Forecasting\dataset\amazon_delivery.csv.xlsx"
)

print("Dataset loaded successfully")
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="Delivery_Time", bins=30, kde=True)

plt.title("Delivery Time Distribution")
plt.xlabel("Delivery Time")
plt.ylabel("Number of Orders")

plt.savefig("../graphs/01_delivery_time_distribution.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Weather")

plt.title("Orders by Weather Condition")
plt.xlabel("Weather")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)

plt.savefig("../graphs/02_weather_distribution.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Traffic",
    y="Delivery_Time",
    showfliers=False
)

plt.title("Delivery Time by Traffic Condition")
plt.xlabel("Traffic Condition")
plt.ylabel("Delivery Time")

plt.savefig("../graphs/06_delivery_time_by_traffic.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="Weather", y="Delivery_Time")

plt.title("Delivery Time by Weather")
plt.xlabel("Weather Condition")
plt.ylabel("Delivery Time")
plt.xticks(rotation=45)
plt.savefig("../graphs/07_delivery_time_by_weather.png")
plt.show()
```

```python
Q1 = df["Delivery_Time"].quantile(0.25)
Q3 = df["Delivery_Time"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df["Delivery_Time"] < lower_limit) |
    (df["Delivery_Time"] > upper_limit)
]

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Number of Outliers:", len(outliers))
print(outliers["Delivery_Time"].describe())
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Weather",
    y="Delivery_Time",
    showfliers=False
)

plt.title("Delivery Time by Weather (Outliers Hidden)")
plt.xlabel("Weather Condition")
plt.ylabel("Delivery Time")

plt.xticks(rotation=45)
plt.show()
```
